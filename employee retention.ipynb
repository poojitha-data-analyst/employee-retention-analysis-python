```python
d={"No":0,"Yes":1}
df["Attrition"]=df["Attrition"].map(d)
df
```

```python
df_corr=df.corr(numeric_only=True)
df_corr
df_corr = df_corr.drop(columns=["StandardHours"], errors="ignore")
df_corr = df_corr.drop(index=["StandardHours"], errors="ignore")
```

```python
%matplotlib inline
sns.heatmap(df_corr,cmap="PiYG",vmin=-1, vmax=1,)
plt.show()
```

```python
df.groupby("JobLevel")['Attrition'].mean()
```

```python
sns.regplot(x="JobLevel",y="Attrition",data=df,logistic=True)
plt.xlabel("Job Level")
plt.ylabel("Attrition Probability")
plt.title("Attrition vs. Job Level")
```

```python
# Group by Department and calculate the count of employees with Attrition = 'Yes'
attrition_by_department = df[df['Attrition'] == 1].groupby('Department').size()

# Display the result
print(attrition_by_department)
```

```python
attrition_by_dept = df[df['Attrition'] == 1].groupby('Department').size()
sns.barplot(x=attrition_by_dept.index, y=attrition_by_dept.values)
plt.xlabel("Department")
plt.ylabel("Number of Employees with Attrition")
plt.title("Attrition by Department")
plt.show()
```
